## Section One

```python
# Importing necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Loading the WooCommerce product export data into a DataFrame
df = pd.read_csv('data/woocommerce-product-export.csv')
```

```python
# Showing a concise summary of the columns using the info() method.
df.info()
```

```python
# Show a summary of statistics pertaining to the columns
df.describe()
```

```python
# Print the first five rows by default.
print(df.head())
```

```python
# Print the last five rows by default.
print(df.tail())
```

```python
# Print  the "total_profit" and “month_number” columns only.
print(df[['total_profit', 'month_number']])
```

```python
# Generating a bar plot by using total profit of all months

plt.bar(df['month_number'], df['total_profit'])
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.title('Company profit per month.')
plt.grid(True, linestyle='--', linewidth=0.9)
plt.show()
```

```python
# Generating a line plot by using total profit of all months

plt.plot(
    df['month_number'],
    df['total_profit'],
    marker='o',
    linestyle='--',
    color='red',
    linewidth=3,
    markeredgecolor='black',
    markerfacecolor='black',
    label='Profit data of last year'
)
plt.xlabel('Month Number')
plt.ylabel('Total Profit')
plt.title('Company profit per month')
plt.xticks(df['month_number'])
plt.legend(loc='lower right')
plt.show()
```

```python
# Creating a multiline plot by using all product sales per months(for loops)
products = {
    'facecream': ('blue', 'Face cream Sales Data'),
    'facewash': ('orange', 'Face Wash Sales Data'),
    'toothpaste': ('green', 'Toothpaste Sales Data'),
    'bathingsoap': ('red', 'Bathingsoap Sales Data'),
    'shampoo': ('purple', 'Shampoo Sales Data'),
    'moisturizer': ('brown', 'Moisturizer Sales Data')
}

for col, (color, label) in products.items():
    plt.plot(df['month_number'], df[col], marker='o', linestyle='-', color=color, label=label)

plt.xlabel('Month Number')
plt.ylabel('Sales units in number')
plt.title('Sales Data')
plt.xticks(df['month_number'])
plt.legend(loc='upper left')
plt.show()
```

```python
# Creating scatter plot for bathingsoap sales data for each month

plt.scatter(
    df['month_number'],
    df['bathingsoap'],
    color='blue',
    label='Bathingsoap Sales Data'
)
plt.xlabel('Month Number')
plt.ylabel('Number of units Sold')
plt.title('bathingsoap Sales Data')
plt.xticks(df['month_number'])
plt.legend(loc='upper left')
plt.grid(True, linestyle='--')
plt.show()
```

### Dataset One:

```python
date=["25/12","26/12","27/12"]
temp=[8.5,10.5,6.8]
```

```python
# Line chart of Date vs Temperature

plt.plot(
    date,
    temp,
    marker='D',
    linestyle='--',
    color='purple',
    label='Temperature Data'
)
plt.xlabel('Date')
plt.ylabel('Temperature')
plt.title('Date-wise Temperature')
plt.xticks(rotation=45)
plt.legend(loc='upper left')
plt.grid(True, linestyle='--')
plt.show()
```

### Dataset two:

```python
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]

```

```python
# Line plot chart of Height vs Weight

plt.plot(
    weight,
    height,
    marker='o',
    markersize=10,
    markerfacecolor='green',
    linestyle='-.',
    color='green',
    label='Weight Data'
)
plt.xlabel('Weight in kg')
plt.ylabel('Height in cm')
plt.title('Average weight with respect to average height')
plt.legend(loc='lower right')
plt.show()
```
